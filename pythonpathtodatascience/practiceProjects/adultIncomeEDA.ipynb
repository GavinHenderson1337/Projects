## 17. Replace Salary Values ['<=50k ' >50K'] With 0 and 1

```python
df.columns
```

```python
df['income'].unique()
```

```python
df['income'].value_counts()
```

```python
sns.countplot(x = 'income', data= df)
```

```python
def income_data(sal):
    if sal == '<=50K':
        return 0
    else:
        return 1
```

```python
df['encoded salary'] =df['income'].apply(income_data)
```

```python
df.head()
```
