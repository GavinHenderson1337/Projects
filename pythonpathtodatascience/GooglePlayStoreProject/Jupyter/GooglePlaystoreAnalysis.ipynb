# 4. Finding and Replacing Nulls

```python
null_counts  = playstore_cleaned.isnull().sum()

print(null_counts)
```

```python
# Create a bar plot to visualize Null Values
plt.figure(figsize=(12, 6))
null_counts[null_counts > 0].plot(kind='bar', color='skyblue')
plt.title('Count of Null Values in Each Column')
plt.xlabel('Columns')
plt.ylabel('Count of Nulls')
plt.xticks(rotation= 45)
plt.show()
```

Here we can see that most columns do not have null values. We can address how we want to handle these nulls in many ways. In this case we are going to drop the nulll values

```python
# Assuming playstore_cleaned is your DataFrame
playstore_cleaned = playstore_cleaned.dropna()

# Optional: If you want to reset the index after dropping rows
playstore_cleaned.reset_index(drop=True, inplace=True)

# Display the cleaned DataFrame
print(playstore_cleaned)
```

## Diving into the data

```python
playstore_cleaned.columns
```

```python
playstore_cleaned['App'].nunique()
```

```python
playstore_cleaned['Category'].nunique()
```

```python
# Identify non-unique apps
duplicate_apps = playstore_cleaned[playstore_cleaned.duplicated(subset='App', keep=False)]

# Drop non-unique apps
playstore_cleaned = playstore_cleaned[~playstore_cleaned['App'].isin(duplicate_apps['App'])]

# Check the cleaned DataFrame
print(playstore_cleaned)
```

### Cleaning Data

```python
# Columns to clean 
cols = ['Installs', 'Price']

# Characters to remove
chars = ['+', ",", "$"]

# Cleaning the specified columns
for col in cols:
    for char in chars:
        # Convert values to string and replace unnecessary characters with an empty string
        playstore_cleaned[col] = playstore_cleaned[col].astype(str).apply(lambda c: c.replace(char, ''))

# Convert the cleaned columns back to appropriate data types
playstore_cleaned['Installs'] = playstore_cleaned['Installs'].astype(int)
playstore_cleaned['Price'] = playstore_cleaned['Price'].astype(float)

# Let's check a random sample of the head of 5 rows
print(playstore_cleaned.head())
```

```python
playstore_cleaned.shape
```

# Category Analysis

```python
category_counts = playstore_cleaned['Category'].value_counts()
```

```python
# Create a bar plot
plt.figure(figsize=(15, 6))
sns.barplot(x=category_counts.index, y=category_counts.values)
plt.xticks(rotation=90)
plt.title('Number of Apps by Category')
plt.xlabel('Category')
plt.ylabel('Number of Apps')
plt.tight_layout()
plt.show()
```

```python
# Drop rows where the 'Category' column is equal to 1.9
playstore_cleaned = playstore_cleaned[playstore_cleaned['Category'] != '1.9']

# Display the updated DataFrame
print(playstore_cleaned)
```

```python
# Group by 'Category', calculate the mean of 'Rating', and reset the index
mean_ratings_by_category = playstore_cleaned.groupby('Category')['Rating'].agg('mean').reset_index()

# Sort the result by mean ratings in descending order
sorted_mean_ratings = mean_ratings_by_category.sort_values(by='Rating', ascending=False)

# Display the sorted result
print(sorted_mean_ratings)
```

```python
plt.figure(figsize=(12, 8))
sns.boxplot(data=playstore_cleaned, y='Category', x='Rating', palette='viridis')

# Add titles and labels
plt.title('Distribution of Ratings by App Category', fontsize=16)
plt.xlabel('Rating', fontsize=14)
plt.ylabel('App Category', fontsize=14)

# Show the plot
plt.show()
```

## Analyzing Free Vs. Paid Apps

```python

# Filter data for 'Free' and 'Paid' apps
free_apps = playstore_cleaned[playstore_cleaned['Type'] == 'Free']
paid_apps = playstore_cleaned[playstore_cleaned['Type'] == 'Paid']

# Combine the data for plotting
free_apps['Type'] = 'Free'
paid_apps['Type'] = 'Paid'
combined_data = pd.concat([free_apps, paid_apps], ignore_index=True)

# Set up the matplotlib figure
plt.figure(figsize=(14, 6))

# Create the first subplot for Ratings
plt.subplot(1, 2, 1)
sns.boxplot(x='Type', y='Rating', data=combined_data)
plt.title('App Ratings by Type')
plt.ylabel('Rating')
plt.xlabel('Type')

# Create the second subplot for Downloads
plt.subplot(1, 2, 2)
sns.boxplot(x='Type', y='Installs', data=combined_data)
plt.title('Downloads by Type')
plt.ylabel('Number of Downloads (Installs)')
plt.xlabel('Type')

# Show the plot
plt.tight_layout()
plt.show()
```
