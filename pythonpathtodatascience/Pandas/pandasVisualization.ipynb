# Pandas Visualization

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r'datasets/Ice Cream Ratings.csv')
df.set_index('Date')
```

Styling -- Use this below

```python
print(plt.style.available)
plt.style.use('dark_background')
```

```python
df.plot(kind= 'line', title='Ice Cream Ratings', subplots= True, xlabel= 'Daily Rating', ylabel= 'Scores')   # The default kind is line plot
                                        # Subplots breaks it into 3 plots
```

```python
df.plot(kind='bar', stacked= True)  # Default is not stacked
```

We can get specific:

```python
df['Flavor Rating'].plot(kind= 'bar', stacked= True) # Only Flavor Rating Column
```

```python
df.plot.barh(stacked=True)
```

Scatterplot

```python
df.plot.scatter(x = 'Texture Rating', y = 'Overall Rating')
```

```python
df.plot.scatter(x = 'Texture Rating', y = 'Overall Rating', s= 500, c= 'Yellow') # 's' Makes the balls bigger
```

Histograms

```python
df.plot.hist(bins= 20)  # Creates how many tick marks at the bottom
```

Boxplot

```python
df.boxplot( )
```

Area

```python
df.plot.area(figsize = (10, 5)) # figsize determines the size of the function
```

Pie Chart

```python
df.plot.pie(y= 'Flavor Rating', figsize = (15,8))
```
