# Data cleaning in Pandas

```python
import pandas as pd
```

```python
df = pd.read_excel(r'datasets/Customer Call List.xlsx')
df
```

```python
# Remove Duplicates
df = df.drop_duplicates()
df
```

```python
# Dropping columms
df = df.drop(columns = 'Not_Useful_Column')
df
```

```python
# Standardizing Values
df["Last_Name"] = df["Last_Name"].str.lstrip('...')
df
df["Last_Name"] = df["Last_Name"].str.lstrip('/')
df
df["Last_Name"] = df["Last_Name"].str.rstrip('_')
df

## Alternative way
# df["Last_Name"] = df["Last_Name"].str.strip('123._/')

```

```python
# Cleaning Phone Numbers
df['Phone_Number'] =df['Phone_Number'].str.replace('[^a-zA-Z0-9]', '')  ## Replacing anything that is not a number
df
```

We need to convert the Phone NUmbers to strings so that we can use the lambda function 

```python
df['Phone_Number'] = df['Phone_Number'].apply(lambda x: str(x))
df
```

```python
df['Phone_Number'] = df['Phone_Number'].apply(lambda x: x[0:3] + '-' + x[3:6] + '-' + x[6:10])
df
```

```python
# Replacing the N/A with blanks
df['Phone_Number'] = df['Phone_Number'].str.replace('nan--', '')
df['Phone_Number'] = df['Phone_Number'].str.replace('Na--', '')

df
```

```python
# Being more specific with our address values
df[['Street_Address', 'State', 'Zip_Code']] = df['Address'].str.split(',', 2, expand=True)
df
```

```python
# Cleaning Yes and No
df['Paying Customer'] = df['Paying Customer'].str.replace('Yes', 'Y')
df['Paying Customer'] = df['Paying Customer'].str.replace('No', 'N')
df

df['Do_Not_Contact'] = df['Do_Not_Contact'].str.replace('Yes', 'Y')
df['Do_Not_Contact'] = df['Do_Not_Contact'].str.replace('No', 'N')
df
```

Cleaning the blanks and N/A's

```python
df = df.replace('N/a', '')
df = df.replace('NaN', '')  # Does not work
df

df.fillna('')
```

```python
# Stripping all people on the do not contact list
for x in df.index:
    if df.loc[x, 'Do_Not_Contact'] == 'Y':
        df.drop(x, inplace = True)
    
df
```

```python
# Stripping all missing numbers
for x in df.index:
    if df.loc[x, 'Phone_Number'] == '':
        df.drop(x, inplace = True)
    
df
```

```python
df = df.reset_index(drop=True)
df
```
