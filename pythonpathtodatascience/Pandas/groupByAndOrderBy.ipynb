# Group by and Agrregating

```python
import pandas as pd
```

```python
df = pd.read_csv(r"datasets/Flavors.csv")
df
```

Group By is the same syntax as SQL

Have to turn it into a variable because its an object

```python
group_by_frame = df.groupby('Base Flavor')
```

Aggregated all integers

AGGREGATE FUNCTIONS

```python
group_by_frame.mean()   
```

```python
df.groupby('Base Flavor').count()
```

```python
df.groupby('Base Flavor').min()
```

```python
df.groupby('Base Flavor').max()
```

```python
df.groupby('Base Flavor').sum()
```

How to perform multiple aggregations on a single columm

```python
df.groupby('Base Flavor').agg({'Flavor Rating':['mean', 'max', 'count', 'sum']})
```

Performing Aggregate Functions on multiple columns

```python
df.groupby('Base Flavor').agg({'Flavor Rating':['mean', 'max', 'count', 'sum'], 'Texture Rating': ['mean', 'max', 'count', 'sum']})
```

Group By Multiple Columns

```python
df.groupby(['Base Flavor', 'Liked']).mean()
```

```python
df.groupby(['Base Flavor', 'Liked']).agg({'Flavor Rating':['mean', 'max', 'count', 'sum'], 'Texture Rating': ['mean', 'max', 'count', 'sum']})
```

Shortcut

```python
df.groupby('Base Flavor').describe()
```
